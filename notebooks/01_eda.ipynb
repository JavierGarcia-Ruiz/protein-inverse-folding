# CATH 4.2 Benchmark: Exploratory Data Analysis
**Project**: Autoregressive 3D Structure-to-Sequence Inverse Folding
**Dataset**: CATH 4.2 Non-redundant Domain Structures (<40% Sequence Identity)
**Goal**: Explore structural class topology, Ramachandran dihedral distributions, and residue environments.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pif.data.constants import CATH_CLASSES, CATH_COLORS, SS_CLASSES, SS_COLORS

df = pd.read_csv('../data/processed/train.csv')
print(f'Train domain structures: {len(df):,}')
df.head()
```

```python
# Class distribution across CATH architectures
fig, ax = plt.subplots(figsize=(8, 4))
counts = df['cath_class_name'].value_counts()
ax.bar(counts.index, counts.values, color=[CATH_COLORS[i] for i in [1, 2, 3, 4]], edgecolor='black')
ax.set_title('CATH Superfamily Structural Class Distribution')
ax.set_ylabel('Number of Domains')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Secondary structure fractions across folds
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.boxplot(data=df, x='cath_class_name', y='helix_fraction', ax=axes[0], palette='Blues')
axes[0].set_title('Alpha-Helix Fraction')
axes[0].tick_params(axis='x', rotation=20)

sns.boxplot(data=df, x='cath_class_name', y='sheet_fraction', ax=axes[1], palette='Greens')
axes[1].set_title('Beta-Sheet Fraction')
axes[1].tick_params(axis='x', rotation=20)

sns.boxplot(data=df, x='cath_class_name', y='loop_fraction', ax=axes[2], palette='Reds')
axes[2].set_title('Loop / Coil Fraction')
axes[2].tick_params(axis='x', rotation=20)
plt.tight_layout()
plt.show()
```
