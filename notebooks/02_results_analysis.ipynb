# Inverse Protein Folding: Benchmark Results & Generative Sequence Design
**Project**: 3D Backbone Inverse Folding via Message Passing Neural Networks
**Comparison**: FrequencyBaseline vs SequenceTransformer vs ProteinMPNN

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

metrics_df = pd.read_csv('../reports/results/metrics_benchmarks.csv')
metrics_df
```

```python
# Plot Sequence Recovery vs Perplexity
fig, ax1 = plt.subplots(figsize=(8, 4.5))
ax2 = ax1.twinx()
x = range(len(metrics_df))

ax1.bar([i - 0.2 for i in x], metrics_df['sequence_recovery'], width=0.4, label='Recovery (%)', color='#009E73')
ax2.bar([i + 0.2 for i in x], metrics_df['perplexity'], width=0.4, label='Perplexity (PPL)', color='#D55E00')

ax1.set_xticks(x)
ax1.set_xticklabels(metrics_df['model'])
ax1.set_ylabel('Sequence Recovery (%)', color='#009E73')
ax2.set_ylabel('Perplexity (PPL)', color='#D55E00')
plt.title('Inverse Folding Model Comparison on Held-Out Test Folds')
plt.tight_layout()
plt.show()
```
